## 2. Calculating NDVI
The NDVI is calculated using the Near-Infrared (B8) and Red (B4) bands:

$$NDVI = \frac{NIR - Red}{NIR + Red}$$

Here is a mock calculation based on a sample array:

```python
nir = np.array([0.8, 0.75, 0.9, 0.85])
red = np.array([0.1, 0.12, 0.11, 0.15])
ndvi = (nir - red) / (nir + red)
f"Mean NDVI for selected ROI: {np.mean(ndvi):.3f}"
```
