# Analyzing Satellite Imagery

Geospatial engineering involves processing massive amounts of orbital data. In this notebook, we explore how to calculate NDVI (Normalized Difference Vegetation Index) from Sentinel-2 bands.

```python
import numpy as np
import rasterio
print("Successfully imported rasterio and numpy")
```

NDVI is calculated using the formula:

$$ NDVI = \frac{NIR - RED}{NIR + RED} $$

Where NIR is Band 8 and RED is Band 4 for Sentinel-2.
