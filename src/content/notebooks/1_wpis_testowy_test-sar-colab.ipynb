## Założenia Analizy
Wykorzystujemy wsteczne rozpraszanie sygnału radarowego (backscatter) z satelity **Sentinel-1**.
Poniżej symulujemy prosty pipeline obliczeniowy.


```python
import numpy as np

# Symulacja danych SAR z regionu rolniczego (wartości w dB)
sigma_naught_db = np.random.uniform(low=-25.0, high=-5.0, size=(5, 5))
threshold_water = -18.0

# Klasyfikacja
water_pixels = np.sum(sigma_naught_db < threshold_water)
print(f"Zainicjowano potok danych z Colaba.")
print(f"Wykryto {water_pixels} pikseli sklasyfikowanych jako woda powierzchniowa.")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.imshow(sigma_naught_db, cmap='viridis')
plt.colorbar(label='Rozproszenie wsteczne (dB)')
plt.title("Symulowana matryca Sentinel-1 (VV)")
plt.show()
```
